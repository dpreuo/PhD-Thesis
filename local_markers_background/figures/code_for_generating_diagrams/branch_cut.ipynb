```python
import numpy as np
import matplotlib.pyplot as plt

def f(k):
    func_of_k = k * np.exp(0.1j * k + 0.001*k**0.8) + (0.4 + 0.7j)*np.sin(k.real) + 1j*np.cos(k.imag + 0.2)
    return np.angle(func_of_k)

k_vals = np.linspace(-10, 10, 1000)
KX, KY = np.meshgrid(k_vals, k_vals)


plt.figure(figsize=(8, 8))
plt.contourf(KX, KY, f(KX + 1j * KY), 100, cmap='RdBu')
plt.colorbar()
plt.savefig('branch_cut.pdf')
```
