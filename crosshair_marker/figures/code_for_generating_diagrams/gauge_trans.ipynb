```python

import numpy as np
from matplotlib import pyplot as plt
from matplotlib import cm
import matplotlib
from mpl_toolkits.mplot3d import Axes3D
matplotlib.rcParams.update({'font.size': 10, 'text.usetex': True, 'font.family': 'serif', 'font.serif': ['Computer Modern'], 'figure.figsize': (10,10)})
plt.rcParams["figure.autolayout"] = True
```

```python
V = 1
nx = 3
ny = 10000
x = np.linspace(0,1,nx)
y = np.linspace(0,1,ny)
X,Y = np.meshgrid(x,y)

Z = V*1/(np.exp((Y - 0.5)*200) + 1)
Z2 = V*(np.exp(-10000*(Y - 0.5)**2))


fig, axes = plt.subplots(1,2,figsize=(8,4), dpi = 300, subplot_kw={'projection': '3d'})
ax, ax2 = axes



# Plot the surface.
ax.get_proj = lambda: np.dot(Axes3D.get_proj(ax), np.diag([1, 1, 0.3, 1]))

ax.plot_surface(X, Y, Z,rstride=1, cstride=1,vmin=-0.2, vmax=1, antialiased = False,cmap = cm.Reds)


# ax.set_zlim(0, 1)               
ax.set_xlabel("x"); ax.set_ylabel("y"); ax.set_zlabel("$V$")

ax.tick_params(label1On=False)
ax.view_init(elev=30., azim=-45)

ax.xaxis.set_pane_color((1.0, 1.0, 1.0, 0.0))
ax.yaxis.set_pane_color((1.0, 1.0, 1.0, 0.0))
ax.xaxis._axinfo["grid"]['color'] =  (1,1,1,0)
ax.yaxis._axinfo["grid"]['color'] =  (1,1,1,0)
ax.zaxis._axinfo["grid"]['color'] =  (1,1,1,0)



ax2.zaxis.set_rotate_label(False)  
ax2.set_xlabel("x"); ax2.set_ylabel("y"); ax2.set_zlabel("$E_y$")
ax2.get_proj = lambda: np.dot(Axes3D.get_proj(ax2), np.diag([1, 1, 0.3, 1]))


ax2.plot_surface(X, Y, Z2,rstride=1, cstride=1, vmin=-0.2, vmax=1, antialiased = False, cmap = cm.Reds)

ax2.tick_params(label1On=False)
ax2.view_init(elev=30., azim=-45)
ax2.xaxis.set_pane_color((1.0, 1.0, 1.0, 0.0))
ax2.yaxis.set_pane_color((1.0, 1.0, 1.0, 0.0))

ax2.xaxis._axinfo["grid"]['color'] =  (1,1,1,0)
ax2.yaxis._axinfo["grid"]['color'] =  (1,1,1,0)
ax2.zaxis._axinfo["grid"]['color'] =  (1,1,1,0)
plt.savefig("gauge_trans.pdf")
```
